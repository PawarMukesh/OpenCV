**Drawing Shapes & Text Over The Images**

```python
""" Create a New Image and
draw the shape over the image"""

import numpy as np
import cv2

```

```python
# Create a white image
img = np.ones((256,256,3))  

cv2.imshow('Image',img)

cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# Create a black image
img = np.zeros((256,256,3))  

cv2.imshow('Image',img)

cv2.waitKey()
cv2.destroyAllWindows()
```

**Drawing Rectangle Over The Image**

```python
# Create a ractangel
img = np.zeros((512,512,3))  

""" 1st parameter for image,
2nd & 3rd parameter for point of ractangle,
4th parameter for color & 5th parameter for thickness """

cv2.rectangle(img, pt1=(50,50), pt2=(200,200), color=(256,0,0), thickness=3) # If you want to fill the whole image with color used -1

cv2.imshow('Image',img)

cv2.waitKey()
cv2.destroyAllWindows()
```

**Drawing Circle Over The Image**

```python
# Create a circle
img = np.zeros((512,512,3))  

""" 1st parameter for image,
2nd parameter for center of the circle, 4th parameter for radius of the circle,
4th parameter for color & 5th parameter for thickness """

cv2.circle(img,center=(100,200),radius=100,color=(0,256,0), thickness=3)

cv2.imshow('Image',img)

cv2.waitKey()
cv2.destroyAllWindows()
```

**Drawing Line Over The Image**

```python
import cv2
import numpy as np
```

```python
# draw line
img = np.zeros((512,512,3))  

""" 1st parameter for image,
2nd & 3rd parameter for point of line,
4th parameter for color & 5th parameter for thickness """

cv2.line(img, pt1=(10,10), pt2=(500,10), color=(0,256,0), thickness=3)

cv2.imshow('Image',img)

cv2.waitKey()
cv2.destroyAllWindows()
```

**Write Text Over The Image**

```python
# Text writing 
img = np.zeros((512,512,3))  

""" 1st parameter for image,
2nd paramter for stating point of text
3rd parameter for font size,
4th parameter for color & 5th parameter for thickness
6th parameter for fontface (Which font you want to use),
6th parameter for text"""

cv2.putText(img,org=(0,200),fontScale=5,text="IMAGE",fontFace=cv2.FONT_HERSHEY_COMPLEX,color=(256,0,0), thickness=3)

cv2.imshow('Image',img)

cv2.waitKey()
cv2.destroyAllWindows()
```

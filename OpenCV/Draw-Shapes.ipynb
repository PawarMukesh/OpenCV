```python
import cv2
import matplotlib.pyplot as plt

# Read Image
image = cv2.imread(r"C:\Users\hp\OneDrive\Pictures\508966-father-love.jpg")

# Convert into RGB
rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Convert into gray scale
gray_scale = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Reduce noise from the image using Gaussian blur
reduce_noise = cv2.GaussianBlur(gray_scale, (5, 5), 0)

# Detect edges after reducing the noise from grayscale image using Canny edge
edge = cv2.Canny(reduce_noise, 80, 100)

# Display the image
fig, axs = plt.subplots(2, 2, figsize=(15, 6))

axs[0, 0].imshow(rgb_image)
axs[0, 0].axis('off')
axs[0, 0].set_title('Original Image')

axs[0, 1].imshow(gray_scale, cmap='gray')
axs[0, 1].axis('off')
axs[0, 1].set_title('Gray Scale Image')

axs[1, 0].imshow(reduce_noise)
axs[1, 0].axis('off')
axs[1, 0].set_title('After Reducing Noise')

axs[1, 1].imshow(edge)
axs[1, 1].axis('off')
axs[1, 1].set_title('Edge Of The Image')

plt.show()
```

```python
# Read the image
img = cv2.imread("your_image_path.jpg")

# Convert the image to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Apply histogram equalization
equalized_image = cv2.equalizeHist(gray_image)

# Display the original and enhanced images side by side
plt.figure(figsize=(10, 5))

# Original Image
plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# Enhanced Image after Histogram Equalization
plt.subplot(1, 2, 2)
plt.imshow(equalized_image, cmap='gray')
plt.title('Enhanced Image (Histogram Equalization)')
plt.axis('off')

plt.show()
```
